```python
# Independent synthetic examples: shuffle indices, keeping feature-target pairs together.
rng = np.random.default_rng(101)
distance = rng.uniform(0.5, 9.5, 150)
z = (distance - 5) / 4.5
duration = 50 + 8*z + 24*z**3 + rng.normal(0, 4, 150)
X_demo = distance.reshape(-1, 1)
y_demo = duration

indices = rng.permutation(len(y_demo))
train_idx, val_idx, test_idx = indices[:90], indices[90:120], indices[120:]
X_train, y_train = X_demo[train_idx], y_demo[train_idx]
X_val, y_val = X_demo[val_idx], y_demo[val_idx]
X_test, y_test = X_demo[test_idx], y_demo[test_idx]
print("Размеры train / validation / test:", len(y_train), len(y_val), len(y_test))

fig, ax = plt.subplots(figsize=(8, 4))
for ids, label, marker in [(train_idx, "Обучение · train", "o"),
                           (val_idx, "Выбор настроек · validation", "s")]:
    sns.scatterplot(x=distance[ids], y=duration[ids], label=label, marker=marker, ax=ax)
ax.set(xlabel="Расстояние, км", ylabel="Время, мин", title="Test отложен: его ответы пока не рассматриваем")
plt.tight_layout()
plt.show()
```

```python
#@title Метрика оценивает результат модели { display-mode: "form" }
display(HTML(intro_viz('<h2>Метрика оценивает результат модели</h2><div class="box def"><div class="t">Метрика / функционал качества · evaluation metric / scoring function</div>Числовой критерий, который оценивает качество работы модели на указанном наборе данных. Для метрики нужно знать направление улучшения: меньше или больше.</div><div class="ml-equation">$$\\begin{aligned}\\operatorname{MAE}_{S}&=\\frac1{n_S}\\sum_{i\\in I_S}|y_i-\\widehat y_i|,\\\\ \\operatorname{MSE}_{S}&=\\frac1{n_S}\\sum_{i\\in I_S}(y_i-\\widehat y_i)^2,\\\\ \\operatorname{RMSE}_{S}&=\\sqrt{\\operatorname{MSE}_{S}}.\\end{aligned}$$</div><div class="where"><b>Где:</b> $S$ — выбранная выборка (train, validation или test); $I_S$ — номера её объектов; $n_S$ — её размер; $y_i$ — ответ; $\\widehat y_i$ — прогноз; MAE — mean absolute error, средняя абсолютная ошибка; MSE — mean squared error, средняя квадратичная ошибка; RMSE — root mean squared error, корень из MSE.</div><div class="ml-table"><table><thead><tr><th>Роль</th><th>Пример</th><th>Как читаем результат</th></tr></thead><tbody><tr><td>Цель обучения · training objective</td><td>Минимизируем MSE на train</td><td>По этой величине изменяем параметры.</td></tr><tr><td>Оценка качества · evaluation</td><td>Считаем MAE на test</td><td>Средняя абсолютная ошибка прогноза в минутах.</td></tr><tr><td>Та же формула, другая роль</td><td>Считаем MSE на test</td><td>Оцениваем уже выбранную модель; веса по test не меняем.</td></tr></tbody></table></div><div class="box idea"><div class="t">Loss и метрика могут совпадать</div>Разница — в роли, а не в обязательном различии формул. Метрики можно отслеживать и во время обучения. MAE и RMSE имеют единицы ответа; MSE — квадрат этих единиц. У всех трёх меньше — лучше. RMSE сильнее реагирует на крупные ошибки, чем MAE.</div><p class="ml-small">«Функционал» здесь означает критерий, который сопоставляет модели число. В литературе так называют и среднюю потерю. Метрика качества не обязана быть метрикой расстояния из линейной алгебры.</p>')))
```

```python
#@title Интерактив — Те же прогнозы — разные оценки ошибки { display-mode: "form" }
display(HTML(intro_viz('<h2>Те же прогнозы — разные оценки ошибки</h2><p>Меняйте прогноз только для третьей доставки. Посмотрите, как меняются его потеря и общие показатели. Здесь ползунок меняет готовый прогноз: это исследование метрик, а не алгоритм обучения.</p><div data-stage></div><div data-readout class="readout" role="status" aria-live="polite"></div><p class="ml-question"><b>Обсудите.</b> Почему одна крупная ошибка сильнее меняет MSE? Как объяснить MAE в минутах заказчику?</p><p class="ml-small">Синтетические данные. Управление графиком не меняет переменные в Python.</p>', widget='loss', data=None)))
```

```python
# The same predictions can be evaluated with different criteria.
y_true_example = np.array([10., 20., 30.])
y_pred_example = np.array([12., 18., 40.])
squared_losses = (y_true_example - y_pred_example)**2
mse = np.mean(squared_losses)
mae = np.mean(np.abs(y_true_example - y_pred_example))
rmse = np.sqrt(mse)
display(pd.DataFrame({"Ответ": y_true_example, "Прогноз": y_pred_example,
                      "Квадратичная потеря": squared_losses}))
print(f"MSE: {mse:.2f} мин²; MAE: {mae:.2f} мин; RMSE: {rmse:.2f} мин")
```

## 8. Параметры и гиперпараметры

```python
#@title Что находит обучение, а что мы выбираем для обучения { display-mode: "form" }
display(HTML(intro_viz('<h2>Что находит обучение, а что мы выбираем для обучения</h2><div class="box def"><div class="t">Гиперпараметры · hyperparameters</div>Настройки семейства моделей или процедуры обучения, которые не определяются тем же шагом подбора параметров. Их задают заранее или выбирают по валидации.</div><div class="ml-equation">$$\\widehat y=w_0+w_1x+\\cdots+w_dx^d.$$</div><div class="where"><b>Где:</b> $x$ — входной признак; $\\widehat y$ — прогноз; $d$ — выбранная степень полинома; $w_0,\\ldots,w_d$ — коэффициенты, которые находим при обучении.</div><div class="ml-table"><table><thead><tr><th>Параметры · parameters</th><th>Гиперпараметры · hyperparameters</th></tr></thead><tbody><tr><td>Значения коэффициентов w₀, …, w_d</td><td>Степень полинома d: какие степени признака доступны модели</td></tr><tr><td>Находятся по train при выбранной настройке</td><td>Задаются до конкретного запуска обучения; могут выбираться по validation</td></tr><tr><td>После обучения определяют прогноз</td><td>Пример настройки обучения: шаг градиентного спуска · learning rate</td></tr></tbody></table></div><div class="box idea"><div class="t">Два уровня выбора</div>Выбрать степень 3 — выбрать форму модели. Найти четыре её коэффициента — обучить модель. Проверять степени по test нельзя: тогда test будет участвовать в выборе.</div><p class="ml-small">В библиотеках параметры конструктора, например <code>degree</code>, часто называются parameters в смысле настроек программы. Здесь различаем их по роли в обучении.</p>')))
```

## 9. Недообучение и переобучение

```python
#@title Малая ошибка на train ещё не означает хорошую модель { display-mode: "form" }
display(HTML(intro_viz('<h2>Малая ошибка на train ещё не означает хорошую модель</h2><div class="ml-cards"><div class="ml-card"><h3>Недообучение</h3><p><b>Underfitting</b></p><p>Модель недостаточно хорошо усвоила закономерность: ошибка велика даже на train. Возможные причины — слишком простая форма или незавершённое обучение.</p></div><div class="ml-card"><h3>Хорошее обобщение</h3><p><b>Good generalization</b></p><p>Модель даёт приемлемую ошибку на новых объектах. Полезный ориентир — качество по сравнению с простым решением.</p></div><div class="ml-card"><h3>Переобучение</h3><p><b>Overfitting</b></p><p>Модель подстроилась под особенности обучающей выборки, которые плохо переносятся на новые данные. Обучающая ошибка выглядит слишком оптимистично.</p></div></div><div class="box def"><div class="t">Сложность модели · model complexity</div>Способность семейства моделей описывать разнообразные зависимости. В нашем эксперименте ею управляет максимальная степень полинома.</div><div class="box idea"><div class="t">Сложная модель не обязательно переобучена</div>Результат зависит также от количества данных, шума и обучения. Высокая ошибка из-за сильного шума сама по себе не означает недообучение. Небольшая разница между train и validation естественна; один разрыв не доказывает переобучение. В нашем эксперименте данные получены одинаково, поэтому удобно исследовать именно влияние сложности.</div>')))
```

```python
#@title Глоссарий: проверка качества { display-mode: "form" }
display(HTML(intro_viz('<h2>Глоссарий: проверка качества</h2><div class="ml-table"><table><thead><tr><th>Термин</th><th>English</th><th>Короткое определение</th></tr></thead><tbody><tr><td>Обучающая выборка</td><td>training set / train</td><td>Данные для подбора параметров модели.</td></tr><tr><td>Валидационная выборка</td><td>validation set / validation</td><td>Данные для сравнения вариантов и выбора настроек.</td></tr><tr><td>Тестовая выборка</td><td>test set / test</td><td>Отложенные данные для итоговой оценки выбранной модели.</td></tr><tr><td>Отложенная выборка</td><td>holdout set</td><td>Данные, не используемые для подбора параметров; нужно уточнять роль.</td></tr><tr><td>Обобщающая способность</td><td>generalization</td><td>Способность хорошо работать на новых объектах в целевых условиях.</td></tr><tr><td>Утечка данных</td><td>data leakage</td><td>Использование информации, которая не должна участвовать в обучении или выборе модели.</td></tr><tr><td>Функция потерь</td><td>loss function</td><td>Критерий ошибки прогноза; в формулах урока ℓ относится к одному объекту.</td></tr><tr><td>Средняя потеря / эмпирический риск</td><td>average loss / empirical risk</td><td>Средняя потеря на рассматриваемой выборке.</td></tr><tr><td>Цель обучения</td><td>training objective</td><td>Критерий, который оптимизирует процедура обучения.</td></tr><tr><td>Метрика / функционал качества</td><td>evaluation metric / scoring function</td><td>Число, оценивающее качество модели на указанном наборе данных.</td></tr><tr><td>Средняя квадратичная ошибка</td><td>mean squared error, MSE</td><td>Среднее квадратов ошибок; меньше — лучше.</td></tr><tr><td>Средняя абсолютная ошибка</td><td>mean absolute error, MAE</td><td>Среднее модулей ошибок в единицах ответа; меньше — лучше.</td></tr><tr><td>Корень из MSE</td><td>root mean squared error, RMSE</td><td>Корень средней квадратичной ошибки в единицах ответа; меньше — лучше.</td></tr><tr><td>Сложность модели</td><td>model complexity</td><td>Способность семейства моделей описывать разнообразные зависимости.</td></tr><tr><td>Недообучение</td><td>underfitting</td><td>Недостаточно хорошее усвоение закономерности, заметное и на обучающих данных.</td></tr><tr><td>Переобучение</td><td>overfitting</td><td>Подстройка под особенности train, которые плохо переносятся на новые данные.</td></tr><tr><td>Базовое решение</td><td>baseline</td><td>Простое решение, относительно которого оценивают другие модели.</td></tr></tbody></table></div>')))
```
