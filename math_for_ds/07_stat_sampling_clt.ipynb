С другой тройкой доставок ответ мог бы получиться другим.

**Новое здесь:** до отбора мы не знаем, какое среднее получим. Оно зависит от случайной выборки.

Дальше программа будет создавать новые вымышленные доставки. **Это уже не выбор из таблицы 200 записей**:
каждое время создаётся заново по одному правилу. У этого учебного источника среднее точно **30 минут**, SD —
**20 минут**. SD (*Standard Deviation*) — знакомое стандартное отклонение, мера разброса значений.
Большинство доставок относительно быстрые, иногда бывают очень долгие.

Готовая функция `make_deliveries(n, rng)` создаёт `n` таких времён. `rng` — генератор случайных чисел.
Его начальное состояние, *seed*, позволяет повторить опыт с теми же числами.

```python
#@title Подготовка учебного источника доставок — выполните один раз { display-mode: "form" }
def make_deliveries(n, rng):
    """Независимые вымышленные доставки: истинное среднее 30 мин, SD 20 мин."""
    return 10 + rng.exponential(scale=20, size=n)

sample_rng = np.random.default_rng(7)
mu_true, sigma_true = 30.0, 20.0
```

```python
#@title Как пользоваться готовым генератором { display-mode: "form" }
display(HTML(lesson7_viz('\n<p>Сейчас важно понимать вход и выход функции: передаём число доставок, получаем массив времён.\nПодробности устройства источника можно пропустить.</p>\n<details><summary>Дополнительно: почему у генератора среднее 30 и SD 20?</summary>\n<p><b>Экспоненциальное распределение (exponential distribution)</b> — непрерывное распределение\nнеотрицательных значений, у которого плотность убывает экспоненциально.\nВызов <code>exponential(scale=20)</code> создаёт значения со средним 20 и SD 20.\nПрибавление 10 сдвигает среднее до 30 и не меняет SD. Это выбранная учебная модель, а не установленный\nзакон реальных доставок.</p>\n<p>Источник: <a href="https://numpy.org/doc/stable/reference/random/generated/numpy.random.Generator.exponential.html">NumPy — Generator.exponential</a>.</p></details>')))
```

**Опыт в Python.** Запустите следующую ячейку пять раз. Генератор продолжает работу, поэтому каждый раз
появляется новая выборка. Саму ячейку подготовки выше повторно запускать не нужно.

```python
n = 20
delivery_times = make_deliveries(n, sample_rng)
sample_mean = delivery_times.mean()

print("Первые пять доставок:", delivery_times[:5].round(1))
print(f"Среднее по {n} доставкам: {sample_mean:.1f} мин")
print(f"Известное в симуляции среднее: {mu_true:.1f} мин")
```

## 4. Насколько различаются наши ответы? Стандартная ошибка

В исходных данных одна доставка могла занять 15 минут, другая — 50 минут. Средние по 100 доставкам
обычно различаются гораздо меньше. Это **два разных разброса**.

| Вопрос | Чем описываем разброс? |
|---|---|
| Насколько различается время отдельных доставок? | Стандартным отклонением времени доставки — SD |
| Насколько различаются средние по выборкам одинакового размера? | Стандартной ошибкой среднего — SE |

## 8. Возвращаемся к автобусам: маленькая SE не означает одинаковые поездки

В практикуме «Моссовет → Азия Молл» мы уже посчитали среднее и SD времени поездок №118 и №169.
Возьмём **те же реальные сводные результаты** из сохранённого вывода задания 1.2.

Утром у №118 было 506 поездок, среднее около 15,3 минуты, SD около 3,8 минуты.
Что получится, если оценить SE этого среднего по формуле из раздела 4?

Сначала посмотрим на расчёт **при предположении независимых поездок**, затем обсудим, насколько
это предположение подходит. В таблице `n` — число поездок, `mean` — среднее, `std` — SD;
`SE_if_independent` — расчётная SE при независимости.

```python
# Реальные округлённые итоги из вывода задания 1.2 предыдущего практикума.
# std — выборочное SD pandas (ddof=1), n — число поездок.
bus_summary = pd.DataFrame([
    ["07–10", 118, 506, 15.3, 3.8],
    ["07–10", 169, 378, 14.7, 5.4],
    ["10–16", 118, 890, 19.0, 4.2],
    ["10–16", 169, 668, 17.3, 5.3],
    ["16–20", 118, 499, 23.2, 5.6],
    ["16–20", 169, 402, 25.1, 8.0],
], columns=["window", "route", "n", "mean", "std"])
bus_summary["SE_if_independent"] = bus_summary["std"] / np.sqrt(bus_summary["n"])
display(bus_summary.round(3))

fig, axes = plt.subplots(1, 2, figsize=(13, 4.8), sharex=True, sharey=True)
labels = [f"{r.window} · №{r.route}" for r in bus_summary.itertuples()]
y = np.arange(len(bus_summary))
for ax, width, title in zip(axes, ["std", "SE_if_independent"],
                          ["Среднее ± 1 SD поездок", "Среднее ± 1 SE при независимости"]):
    ax.errorbar(bus_summary["mean"], y, xerr=bus_summary[width], fmt="o", capsize=5)
    ax.set(title=title, xlabel="Время в пути, мин", yticks=y, yticklabels=labels)
axes[0].invert_yaxis()
fig.suptitle("Одинаковая шкала: разброс поездок и неопределённость среднего")
plt.tight_layout()
plt.show()
```

**Как прочитать графики.** Слева полоска показывает одно SD в каждую сторону от среднего.
Справа — одну расчётную SE. Масштаб в минутах одинаковый, поэтому хорошо видно, что это разные величины.
Полоскам здесь не приписывается вероятность 95%; это не результат статистического теста.

Для утреннего №118 расчётная SE получается около **0,17 минуты**. Это мера точности оценки среднего
при наших предположениях. Она не обещает, что следующая поездка отклонится от среднего только на 0,17 минуты.

**Что меняется в реальных данных?** Поездки одного дня могут вместе замедлиться из-за пробок.
Значит, считать все строки независимыми может быть неверно. Формулу мы умеем применять, но её условия
нужно проверить. Как учитывать зависимость, будем разбирать отдельно; ниже есть дополнительная симуляция.
