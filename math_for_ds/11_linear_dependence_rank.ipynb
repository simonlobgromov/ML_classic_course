## 5. Ранг как число направлений после преобразования

```python
#@title Умножение на матрицу собирает комбинацию столбцов { display-mode: "form" }
display(HTML(lesson11_viz('<p>В прошлом занятии мы видели: умножение матрицы на вектор — это линейная комбинация столбцов матрицы.</p><div class=\'d11-equation\'>$$Ax=x_1c_1+\\cdots+x_nc_n.$$</div><div class=\'where\'><b>Обозначения.</b> $A$ — матрица с $n$ столбцами; $c_1,\\ldots,c_n$ — её столбцы; $x$ — входной вектор из $n$ координат; $x_i$ — его $i$-я координата; $Ax$ — выходной вектор.</div><div class="box idea"><div class="t">Геометрический смысл ранга</div>Ранг — число независимых направлений среди всех возможных выходных векторов. Для матрицы 2 × 2: ранг 2 — выходы заполняют плоскость; ранг 1 — лежат на прямой через начало координат; ранг 0 — все совпадают с нулевым вектором.</div><div class="box def"><div class="t">Вырожденная матрица · singular matrix</div><p><b>Квадратная матрица называется вырожденной, если её ранг меньше её порядка.</b> Порядок квадратной матрицы — число её строк, равное числу столбцов. Если ранг равен порядку, матрица невырожденная.</p><div class=\'d11-equation\'>$$A\\in\\mathbb R^{n\\times n}:\\qquad A\\text{ вырождена}\\ \\Longleftrightarrow\\ \\operatorname{rank}(A)<n.$$</div><div class=\'where\'><b>Обозначения.</b> $A$ — квадратная матрица; $n$ — её порядок; $\\mathbb R$ — вещественные числа; $\\operatorname{rank}(A)$ — ранг; $\\Longleftrightarrow$ означает «тогда и только тогда».</div></div><p>Для прямоугольных матриц здесь используем слова «полный» и «неполный ранг».</p>')))
```

```python
#@title Интерактив — Плоскость, прямая или точка { display-mode: "form" }
display(HTML(lesson11_viz('<div class="widget"><div class="widget-title">Плоскость, прямая или точка</div><p class="widget-sub">Переключайте три матрицы. Слева — исходные точки, справа — результаты умножения на матрицу. Особенно проследите за двумя разными входами u = (1, 0) и v = (0, 1).</p><div data-stage></div><div class="readout" data-readout role="status" aria-live="polite"></div><p class="d11-question"><b>Проверь себя.</b> При каком преобразовании u и v дают один и тот же результат? Сможем ли мы по такому результату однозначно узнать, какой из этих двух входов использовался?</p><p class="d11-footnote">Интерактивная иллюстрация. Изменения здесь не меняют переменные в Python.</p></div>', widget='collapse', data=None)))
```

```python
A = np.array([[1, 1], [2, 2]])
u = np.array([1, 0])
v = np.array([0, 1])
print("Ранг:", np.linalg.matrix_rank(A))
print("Разные входы:", u, "и", v)
print("Выходы:", A @ u, "и", A @ v)
print("Выходы совпали:", np.array_equal(A @ u, A @ v))
```

```python
#@title Почему теряется возможность различить входы { display-mode: "form" }
display(HTML(lesson11_viz('<p>Столбцы этой матрицы одинаковы. Поэтому два разных входа дали один результат. Зная только результат, нельзя определить, какой именно вход был использован.</p><div class="box idea"><div class="t">Вырожденность и потеря информации</div>У вырожденной квадратной матрицы обязательно найдутся разные входные векторы с одинаковым выходом. Это свойство преобразования, а не ошибка вычислений.</div>')))
```

```python
# Observations are rows: multiply by A.T to transform each point.
grid = np.arange(-2, 3)
points = np.array([(x, y) for x in grid for y in grid])
transformed = points @ A.T
colors = points[:, 0] + 2 * points[:, 1]
fig, axes = plt.subplots(1, 2, figsize=(10, 4.5))
for ax, data, title in zip(axes, [points, transformed],
                           ["Входы: часть плоскости", "Выходы: точки на прямой"]):
    ax.scatter(data[:, 0], data[:, 1], c=colors, cmap="viridis", s=65)
    ax.axhline(0, color="gray", lw=0.8)
    ax.axvline(0, color="gray", lw=0.8)
    ax.set(xlim=(-9, 9), ylim=(-9, 9), xlabel="Первая координата",
           ylabel="Вторая координата", title=title)
    ax.set_aspect("equal")
    ax.grid(alpha=0.25)
plt.tight_layout()
plt.show()
```
