## 4. CDF — кумулятивная функция распределения (Cumulative Distribution Function)

```python
#@title CDF и PDF — накопленная вероятность и плотность { display-mode: "form" }
display(HTML(lesson6_viz('\n<p>До этого мы рассматривали <b>PDF — Probability Density Function</b>, функцию плотности вероятности:\nвероятность интервала вычислялась площадью под её графиком. Теперь вводим другую функцию —\n<b>CDF — Cumulative Distribution Function</b>, которая сразу возвращает накопленную вероятность.</p>\n<p><b>Cumulative</b> — кумулятивная, то есть накопленная; <b>Distribution</b> — распределение;\n<b>Function</b> — функция. В русскоязычной математической литературе короткое название\n«функция распределения» обычно означает именно CDF. В этой тетради будем явно писать <b>CDF</b>,\nчтобы отличать её от <b>PDF</b> и от закона распределения как общего объекта.</p>\n<div class="box def"><div class="t">Определение — CDF, кумулятивная функция распределения</div><b>CDF (Cumulative Distribution Function)</b> случайной величины — функция, которая каждому вещественному порогу ставит в соответствие вероятность получить значение меньше либо равное этому порогу. Она накапливает вероятность всех значений до порога включительно, существует для любой вещественной случайной величины и однозначно задаёт её закон распределения.</div><div class=\'d6-equation\'>$$F_X(x)=P(X\\le x).$$</div><div class=\'where\'><b>Обозначения.</b> $X$ — случайная величина; $x$ — вещественный порог; $P$ — вероятность; $F_X$ — CDF величины $X$; $P(X\\le x)$ — накопленная вероятность до $x$ включительно. Значение $F_X(x)$ — безразмерное число от 0 до 1.</div><div class="box def"><div class="t">Напоминание определения — PDF, функция плотности вероятности</div><b>PDF (Probability Density Function)</b> — неотрицательная измеримая функция, интеграл которой по любому борелевскому множеству равен вероятности попадания в это множество. В частности, вероятность интервала равна площади под PDF на этом интервале, а полная площадь равна 1. Это определение из пункта 3. Значение PDF в отдельной точке — плотность, измеряемая в обратных единицах признака.</div><div class=\'d6-equation\'>$$F_X(x)=\\int_{-\\infty}^{x}f_X(t)\\,dt\\quad\\text{при наличии плотности}.$$</div><div class=\'where\'><b>Обозначения.</b> $F_X$ — CDF; $f_X$ — PDF той же случайной величины $X$; $t$ — переменная интегрирования; $x$ — верхняя граница; $dt$ — элемент ширины. Интеграл — вся площадь под PDF слева от $x$; её величина становится высотой графика CDF в точке $x$.</div><h3>PDF как производная CDF</h3>\n<p>Передвинем порог немного вправо. CDF увеличится ровно на вероятность добавившегося интервала.\nОтношение прироста CDF к ширине интервала показывает, сколько вероятности прибавилось на единицу длины.</p><div class=\'d6-equation\'>$$\\frac{F_X(x+h)-F_X(x)}{h}=\\frac{P(x<X\\le x+h)}{h},\\qquad h>0.$$</div><div class=\'where\'><b>Обозначения.</b> $x$ — исходный порог; $h$ — положительная ширина добавленного интервала; $F_X$ — CDF; $F_X(x+h)-F_X(x)$ — прирост накопленной вероятности; $P(x<X\\le x+h)$ — вероятность добавленного интервала. Деление на $h$ даёт среднюю плотность вероятности на этом интервале.</div><div class="box def"><div class="t">Определение — производная CDF в точке</div>Производная CDF — предел отношения её приращения к приращению аргумента, когда приращение аргумента стремится к нулю, если этот предел существует. Она описывает локальную скорость изменения накопленной вероятности. Штрих и запись d/dx обозначают производную по x.</div><div class=\'d6-equation\'>$$F_X\'(x)=\\frac{dF_X(x)}{dx}=\\lim_{h\\to0}\\frac{F_X(x+h)-F_X(x)}{h}=f_X(x)\\quad\\text{если PDF непрерывна в }x.$$</div><div class=\'where\'><b>Обозначения.</b> $F_X$ — CDF; $F_X\'(x)$ и $dF_X(x)/dx$ — её производная в точке $x$; $h$ — ненулевое приращение аргумента, которое может быть положительным или отрицательным; $\\lim_{h\\to0}$ — предел при стремлении этого приращения к нулю; $f_X$ — PDF. Равенство с PDF здесь сформулировано для точек её непрерывности; у нормальной плотности это каждая точка.</div><p><b>Геометрический смысл:</b> там, где PDF высокая, CDF быстро растёт; там, где PDF мала,\nCDF меняется медленно. В виджете это показывает <b>касательная</b> — прямая через выбранную точку,\nнаклон которой совпадает с локальным наклоном графика CDF. Её наклон, то есть изменение вертикальной\nкоординаты на единицу горизонтальной, равен производной CDF и высоте PDF при том же значении аргумента.</p>\n<p>CDF существует и у Бернулли, но там она имеет скачки. Обычной PDF у дискретного распределения нет:\nего вероятности описываются функцией вероятностей PMF из пункта 1. Равенство PDF и производной CDF\nнельзя переносить на произвольный закон без условий.</p>\n<div class="d6-table"><table><tr><th>Функция</th><th>Что показывает высота графика при x</th><th>Как получить вероятность интервала</th></tr>\n<tr><td>PDF — Probability Density Function</td><td>Плотность вероятности $f_X(x)$</td><td>Площадь под PDF между границами</td></tr>\n<tr><td>CDF — Cumulative Distribution Function</td><td>Накопленную вероятность $F_X(x)$</td><td>Разность значений CDF на границах</td></tr></table></div>\n<p><b>Свойства CDF:</b> её значения лежат между 0 и 1; она не убывает. Она непрерывна справа —\nпри приближении порога к выбранному значению справа значение CDF стремится к значению CDF в этой точке.\nПри уходе порога к минус бесконечности CDF стремится к 0, к плюс бесконечности — к 1.\nУ дискретного распределения величина скачка CDF равна вероятности соответствующего отдельного значения.</p><div class=\'d6-equation\'>$$P(a<X\\le b)=F_X(b)-F_X(a),\\qquad P(X>b)=1-F_X(b).$$</div><div class=\'where\'><b>Обозначения.</b> $X$ — случайная величина; $a\\le b$ — границы; $F_X$ — её CDF; $F_X(b)$ — вероятность до правой границы включительно; $F_X(a)$ — вероятность до левой включительно. Вычитание оставляет интервал, открытый слева. $1-F_X(b)$ — вероятность строго правее $b$. Эти формулы верны для любого закона распределения.</div><p>В моделях с плотностью вероятность отдельной точки нулевая, поэтому включение или\nисключение концов интервала не меняет результат. Для Бернулли такая замена может изменить вероятность.</p>\n<p>В примере стандартной нормали PDF имеет форму колокола, а CDF постепенно накапливает вероятность\nот 0 до 1. Ползунок задаёт одну точку на обеих шкалах.</p>\n<p>Источник определений, английских названий и связи PDF с производной CDF:\n<a href="https://docs.scipy.org/doc/scipy/tutorial/stats/continuous.html" target="_blank" rel="noopener noreferrer">SciPy — Continuous Statistical Distributions</a>.</p>')))
```

## 7. От вероятности к наблюдениям: сколько из 1000 окажутся не ниже 190 см?

В пункте 6 мы нашли площадь правого хвоста. Теперь используем её для прогноза и сравним прогноз с данными.
Событие остаётся прежним: **рост от 190 см**, модель — нормальная, среднее 170 см, стандартное отклонение 10 см.

**Шаг 1. Получаем прогноз до появления выборки.** Вычислим вероятность события, а затем ожидаемое
число таких наблюдений среди 1000. Для расчёта вероятности нужны параметры модели и порог;
сама выборка ещё не нужна.

```python
#@title Как площадь превратить в прогноз количества { display-mode: "form" }
display(HTML(lesson6_viz('<div class="box def"><div class="t">SF — Survival Function, функция выживания</div><p>Здесь это <b>функция вероятности правого хвоста</b>: каждому порогу она ставит в соответствие вероятность получить значение строго выше него. Историческое название связано с задачами о времени жизни; мы применяем ту же функцию к росту.</p></div><div class=\'d6-equation\'>$$S_X(c)=P(X>c)=1-F_X(c),\\qquad p=P(X\\ge190)=S_X(190)\\approx0.02275.$$</div><div class=\'where\'><b>Обозначения.</b> $X$ — случайный рост; $c$ — порог; $S_X$ — SF; $F_X(c)=P(X\\le c)$ — CDF этой же величины; $p$ — вероятность выбранного события по нормальной модели. Для нормального распределения $P(X=c)=0$, поэтому «выше» и «не ниже» имеют одинаковую вероятность.</div><div class="box def"><div class="t">Ожидаемое число — математическое ожидание количества</div><p>Пусть $N$ — случайное количество наблюдений, удовлетворяющих нашему условию, в выборке размера $n$. <b>Математическое ожидание</b> этого количества — среднее возможных значений $N$, взвешенных по их вероятностям. Его можно понимать как среднее количество при очень большом числе повторений опыта.</p></div><div class=\'d6-equation\'>$$\\mathbb E[N]=\\sum_{j=0}^{n}j\\,P(N=j)=np,\\qquad n=1000\\ \\Longrightarrow\\ \\mathbb E[N]\\approx22.75.$$</div><div class=\'where\'><b>Обозначения.</b> $N$ — случайное количество значений не ниже 190 см; $\\mathbb E[N]$ — его математическое ожидание; $j$ — возможное количество от 0 до $n$; $P(N=j)$ — вероятность получить именно $j$ таких значений; $\\sum$ — сумма по всем возможным количествам; $n$ — размер выборки; $p$ — одна и та же вероятность события для каждого наблюдения. В нашем эксперименте наблюдения независимо получаются из одной заданной модели.</div><p><b>22,75 — среднее количество за много повторений.</b> В одной выборке количество целое:\nможет получиться 19, 23, 28 или другое число. Модель не обещает ровно 23 человека в каждой тысяче.</p>\n<p>Источники: <a href="https://docs.scipy.org/doc/scipy/reference/generated/scipy.stats.norm.html">SciPy — norm, методы CDF и SF</a>;\n<a href="https://www.itl.nist.gov/div898/handbook/eda/section3/eda366i.htm">NIST — среднее числа наступлений события в независимых испытаниях</a>.</p>')))
```

В следующей ячейке `norm.sf(...)` считает **ту самую площадь справа от 190 см**, которую мы закрашивали.
`loc=mu` задаёт среднее, `scale=sigma` — стандартное отклонение. `p_model` — вероятность по модели,
`expected_count` — ожидаемое количество. Для правого хвоста `sf` предпочтительнее `1 - cdf`:
в далёких хвостах она позволяет избежать потери численной точности при вычитании близких чисел.

```python
import numpy as np
from scipy.stats import norm

mu, sigma = 170.0, 10.0       # параметры нормальной модели, см
threshold = 190.0            # порог роста, см
n = 1000                    # сколько наблюдений будет в одной выборке

p_model = norm.sf(threshold, loc=mu, scale=sigma)
expected_count = n * p_model

print(f"Событие: рост от {threshold:g} см")
print(f"Вероятность по модели: {p_model:.3%}")
print(f"Ожидаемое количество среди {n}: {expected_count:.2f}")
```

**Шаг 2. Получаем одну выборку и считаем, что произошло.** Создадим 1000 учебных наблюдений
из этой же нормальной модели. Так мы точно знаем закон, породивший данные, и можем изучить
случайные расхождения с прогнозом. Это эксперимент с искусственными данными; он сам по себе
ничего не доказывает о распределении роста реальных людей.

Ячейка ниже создаёт **генератор псевдослучайных чисел** — объект, который выдаёт последовательность
чисел для воспроизводимого моделирования случайных выборок. `42` задаёт начальное состояние
(англ. *seed*). Выполните эту ячейку один раз, затем повторяйте ячейку с наблюдениями под ней.
Если заново создать генератор с тем же seed, последовательность начнётся сначала.
Подробнее: [NumPy — Random Generator](https://numpy.org/doc/stable/reference/random/generator.html).

```python
sample_rng = np.random.default_rng(42)
```

Прочитаем действия до запуска:

1. `normal(...)` создаёт `n` новых значений роста из заданной нормальной модели.
2. `sample >= threshold` проверяет условие для каждого наблюдения и возвращает массив `True`/`False`.
3. `sum()` считает подходящие значения: при суммировании `True` учитывается как 1, `False` — как 0.
4. Деление количества на размер выборки даёт **наблюдаемую, или эмпирическую, долю** `p_sample`.

```python
# Повторяйте эту ячейку: генератор продолжит последовательность и даст новую выборку.
sample = sample_rng.normal(loc=mu, scale=sigma, size=n)
is_tall = sample >= threshold
observed_count = int(is_tall.sum())
p_sample = observed_count / len(sample)

print(f"Событие: рост от {threshold:g} см; наблюдений: {len(sample)}")
print(f"По модели: вероятность {p_model:.3%}; ожидаемое количество {expected_count:.2f}")
print(f"По данным: количество {observed_count}; доля {observed_count}/{len(sample)} = {p_sample:.3%}")
```

**Как прочитать результат опыта.** Когда генератор нормальный, различия между долей и вероятностью
возникают из-за случайного состава выборок. При увеличении размера выборки доля становится устойчивее,
но точного совпадения в каждом опыте ждать не следует. Когда генератор — смесь, доля при большом размере
приближается к вероятности смеси. Большее число наблюдений не исправляет неверно выбранную форму модели.

При пороге 170 см обе симметричные модели дают вероятность 50%. Совпадение вероятностей для одного
порога ещё не означает совпадение распределений.

На реальных данных закон генератора неизвестен. Кроме случайности выборки, расхождение может быть связано
с тем, как собирали данные, как оценивали параметры и подходит ли выбранная модель.
Одно расхождение само по себе не доказывает ненормальность; одно совпадение не доказывает нормальность.

Теперь мы умеем пройти всю цепочку: **сформулировать событие → выбрать площадь → вычислить вероятность
→ получить прогноз количества → посчитать долю в данных → объяснить возможное расхождение**.
