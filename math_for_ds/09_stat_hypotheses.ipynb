## 4. На сколько минут изменилась доставка?

```python
#@title Интервал для разницы средних { display-mode: "form" }
display(HTML(lesson9_viz('<p>P-value оценивает необычность результата при нулевой модели. Теперь оценим <b>величину изменения и её точность</b>.</p><div class="box def"><div class="t">Доверительный интервал · confidence interval</div><p>Интервал со случайными границами, построенный по выборке методом, который покрывает неизвестный параметр с заданной вероятностью.</p></div><p>Для 95%-го метода при многократном повторении исследования около 95% построенных интервалов содержали бы истинный параметр, если условия выполнены. Наш бутстрап-интервал приближённый.</p><div class="box idea"><div class="t">Какой параметр оцениваем</div><p>Строим интервал <b>для истинной разницы средних B − A</b>. Он описывает точность оценки средней разницы, а не время следующего заказа.</p></div>')))
```

```python
#@title Бутстрап двух независимых групп { display-mode: "form" }
display(HTML(lesson9_viz('\n<ol><li>Отберём с возвращением времена внутри A, сохраняя размер группы.</li>\n<li>Независимо сделаем то же внутри B.</li><li>Посчитаем разницу средних. Повторим много раз.</li></ol>\n<p><code>choice</code> выбирает значения, <code>size</code> задаёт число наблюдений,\n<code>replace=True</code> разрешает повторы. Метки A/B сохраняются: группы для отбора не объединяем.</p>\n')))
```

```python
# Worked example: separate bootstrap resampling within the two groups.
rng_boot = np.random.default_rng(9104)
B = 9999
boot_diffs = []
for _ in range(B):
    a_star = rng_boot.choice(a, size=len(a), replace=True)
    b_star = rng_boot.choice(b, size=len(b), replace=True)
    boot_diffs.append(b_star.mean() - a_star.mean())
boot_diffs = np.array(boot_diffs)
```

```python
#@title Границы бутстрап-интервала { display-mode: "form" }
display(HTML(lesson9_viz("<div class='d9-equation'>$$I_{95\\%}=\\left[q^*_{0.025};\\ q^*_{0.975}\\right].$$</div><div class='where'><b>Обозначения.</b> $I_{95\\%}$ — процентильный 95%-й бутстрап-интервал для $\\Delta=\\mu_B-\\mu_A$; $q^*_{0.025},q^*_{0.975}$ — квантили 2,5% и 97,5% среди бутстрап-разниц средних. Звёздочка отмечает величины, полученные бутстрапом.</div>")))
```

```python
ci = np.quantile(boot_diffs, [0.025, 0.975])
print(f'Оценка разницы: {observed:.2f} мин; 95%-й бутстрап-ДИ: [{ci[0]:.2f}; {ci[1]:.2f}] мин.')
plot_bootstrap(boot_diffs, observed, ci)
```
